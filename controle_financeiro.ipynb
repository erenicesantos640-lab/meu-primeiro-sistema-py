```python
print("=== CALCULADORA DE IMC ===")

nome = input("Seu nome: ")
peso = float(input("Seu peso (ex: 70.5): "))
altura = float(input("Sua altura (ex: 1.68): "))

imc = peso / (altura * altura)

print("\n", nome, "seu IMC é:", round(imc, 2))

if imc < 18.5:
  print("Abaixo do peso")
elif imc < 24.9:
  print("Peso normal - Parabéns! 👏")
elif imc < 29.9:
  print("Sobrepeso")
else:
  print("Obesidade")
```

```python
print("🛒 MERCADINHO DA ERENICE")

produtos = ["cafe", "pão", "leite", "açucar"]
precos = [5.50, 3.00, 4.20, 6.00]

while True:
  nome = input("\nNome do produto (ou 'fim' para encerrar): ")

  if nome == "fim":
    break

  preco = float(input(f"Preço do {nome}: R$ "))

  produtos.append(nome)
  precos.append(preco)
  print(f"{nome} adicionado!")

print("\n=== NOTA FISCAL ===")
total = 0
for i in range(len(produtos)):
  print(f"{produtos[i]} - R$ {precos[i]}")
  total = total + precos[i]

print("-------------------")
print(f"TOTAL: R$ {total}")

if total > 100:
  print("Você gastou muito hoje! 😅")
else:
  print("Compra econômica! 👏")
```

```python
print("Calculadora que não quebra")

while True:
  try:
    num1 = float(input("\nDigite o primeiro número: "))
    num2 = float(input("Digite o segundo número: "))

    divisao = num1 / num2
    print(f"Resultado da divisão: {divisao}")
    break # se deu certo, sai do loop

  except ValueError:
    print("Ops! Você digitou letra em vez de número. Tenta de novo!")

  except ZeroDivisionError:
    print("Não pode dividir por zero! Tenta outro número.")

print("\nPrograma terminou sem travar! 👏")
```

```python
def mostrar_menu():
  print("\n=== MEU CONTROLE FINANCEIRO ===")
  print("1 - Adicionar gasto")
  print("2 - Ver todos os gastos")
  print("3 - Ver total gasto")
  print("4 - Sair")

gastos = [] # lista de dicionários

while True:
  mostrar_menu()
  opcao = input("Escolha uma opção: ")

  if opcao == "1":
    try:
      nome = input("O que você comprou? ")
      valor = float(input(f"Quanto custou {nome}? R$ "))
      gastos.append({"item": nome, "valor": valor})
      print(f"{nome} adicionado com sucesso!")
    except ValueError:
      print("Valor inválido! Use ponto, ex: 10.50")

  elif opcao == "2":
    if len(gastos) == 0:
      print("Nenhum gasto ainda!")
    else:
      print("\n--- SEUS GASTOS ---")
      for g in gastos:
        print(f"- {g['item']}: R$ {g['valor']}")

  elif opcao == "3":
    total = 0
    for g in gastos:
      total += g["valor"]
    print(f"\n💰 TOTAL GASTO: R$ {total}")

    if total > 500:
      print("Alerta: gastou bastante esse mês!")
    else:
      print("Você está controlando bem!")

  elif opcao == "4":
    print("Até amanhã, programadora Erenice! 👋")
    break

  else:
    print("Opção inválida! Escolha 1, 2, 3 ou 4")
```
